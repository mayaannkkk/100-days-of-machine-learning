```python
import pandas as pd
import numpy as np
import random

df = pd.DataFrame(
                  [
                      [6.82, 118, 0],
                      [6.36, 125, 1],
                      [5.39, 99, 1],
                      [5.50, 106, 1],
                      [6.39, 148, 0],
                      [9.13, 148, 1],
                      [7.17, 147, 1],
                      [7.72, 72, 0]
                   ], columns = ['cgpa',	'iq',	'is_placed']
)

df
```

```python
df['log_odds'] = np.log(5/3)
df
```

```python
df['pre1(probality)'] = 1 / (1+np.exp(-np.log(5/3)))
df
```

```python
df['resd1'] = df['is_placed'] - df['pre1(probality)']
```

```python
from sklearn.tree import DecisionTreeRegressor, plot_tree
```

```python
dt = DecisionTreeRegressor(max_leaf_nodes=3)
dt.fit(df.iloc[:,0:2],df.iloc[:,-1])
```

```python
plot_tree(dt)
```

```python
df['leaf_entry1'] = dt.apply(df.iloc[:,0:2])
df
```

```python
def return_logs(leaf):
  temp_df = df[df['leaf_entry1'] == leaf]
  num = temp_df['resd1'].sum()

  den = sum(temp_df['pre1(probality)'] * (1 - temp_df['pre1(probality)']))
  return round(num/den,2)
```

```python
df['pre2(log_odds)'] = df['log_odds'] + df['leaf_entry1'].apply(return_logs)
```

```python
df
```

```python
df['prob2(probability)'] = 1 / (1 + np.exp(-df['pre2(log_odds)']))
df
```

```python
df['res2'] = df['is_placed'] - df['prob2(probability)']
df
```

```python
dt1 = DecisionTreeRegressor(max_leaf_nodes=3)
dt1.fit(df.iloc[:,0:2],df.iloc[:,-1])
```

```python
plot_tree(dt1)
```

```python
df['leaf_entry2'] = dt1.apply(df.iloc[:,0:2])
df
```

```python
def return_logs(leaf):
  num = df[df['leaf_entry2'] == leaf]['res2'].sum()
  den = sum(df[df['leaf_entry2'] == leaf]['prob2(probability)'] * (1 - df[df['leaf_entry2'] == leaf]['prob2(probability)']))
  return round(num/den,2)
```

```python
df['pre3(log_odds)'] = df['log_odds'] + df['pre2(log_odds)'] + df['leaf_entry2'].apply(return_logs)
df
```

```python
df['prob3(probability)'] = 1 / (1 + (np.exp(-df['pre3(log_odds)'])))
```

```python
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import train_test_split
X, y = load_breast_cancer(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.3, random_state = 42)

model = GradientBoostingClassifier()
model.fit(X_train, y_train)

pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(pred,y_test)
```
