```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeRegressor, plot_tree
```

```python
df = pd.DataFrame()
```

```python
df['cgpa'] = [90,100,110,120,80]
df['iq'] = [8,7,6,9,5]
df['salary'] = [3,4,8,6,3]
```

```python
df
```

```python
df['pred1'] =np.mean(df['salary'])
df['resd1'] = df['salary'] - df['pred1']
df
```

```python
X = df.iloc[:,0:2]
y = df.iloc[:,-1]
dt = DecisionTreeRegressor()
dt.fit(X,y)
```

```python
plot_tree(dt)
```

```python
df['pred2'] = dt.predict(X)
```

```python
df
```

```python
df['resd2'] = df['salary'] - (df['pred1'] + 0.1 * df['pred2'])
```

```python
df
```

```python
dt1 = DecisionTreeRegressor()
dt1.fit(X,df['resd2'])
```

```python
plot_tree(dt1)
```

```python
df['pred3'] = dt1.predict(X)
df
```

```python
df['y_pred_model2'] = df['pred1'] - 0.1 * df['pred2']
```

```python
df
```

```python
df['resd3'] = df['salary'] - (df['pred1'] + 0.1 * df['pred2'] + 0.1 * df['pred3'])
df
```
