```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv("weight-height.csv")
```

```python
df.head()
```

```python
sns.boxplot(df['Height'])
```

```python
upper = df['Height'].quantile(0.99)
lower = df['Height'].quantile(0.01)
```

```python
new_df= df[(df['Height'] <= upper) & (df['Height'] >= lower)]
new_df
```

Capping

```python
df["Height"]= np.where(df['Height'] > upper, upper,
                      np.where(df['Height'] < lower, lower, df['Height']))
```
