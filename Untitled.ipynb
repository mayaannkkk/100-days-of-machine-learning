```python
import pandas as pd 
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv('daily_gym_attendance_workout_data.csv')
```

```python
df.sample(5)
```

```python
df.shape
```

```python
df['gender'].value_counts()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test=train_test_split(df.drop('attendance_status',axis=1),
                                                 df['attendance_status'],
                                                 test_size=0.3,
                                                 random_state=0)
```

```python
X_train.shape
```

```python
transformer = ColumnTransformer(transformers =[ 
                               ('tnf1',SimpleImputer(),['age'],'member_id','visit_date'),
                               ('tnf2',OrdinalEncoder(categories=[['Monthly','Quarterly','Annual']]),['membership_type']),
                               ('tnf3',OneHotEncoder(sparse_output=False),['workout_type','gender']),
],remainder='passthrough')
```

```python
X_train.head()
```

```python
X_train_new=transformer.fit_transform(X_train)
```
