```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("train.csv",usecols=["SibSp","Parch","Age","Survived",'Pclass'])
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
X= df.iloc[:,1:5]
y= df['Survived']
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=20))
```

```python
X['family']= X["SibSp"] + X['Parch']
```

```python
X.head()
```

```python
def my_fun(num):
    if num == 1:
        return 0
    elif num>1 & num<=4:
        return 1
    else: 
        return 2
```

```python
X['family_type'] = X['family'].apply(my_fun) 
```

```python
X.drop(['SibSp','Parch','family'],axis=1)
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=20))
```
